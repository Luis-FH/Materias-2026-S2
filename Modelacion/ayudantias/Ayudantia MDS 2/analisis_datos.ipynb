## Paso 3: Data Wrangling 

Limpiamos, transformamos y organizamos los datos para el analisis.

### 3.1 Valores nulos

| Funcion | Que hace |
|---------|----------|
| `.isnull().sum()` | Cuenta nulos por columna |
| `.dropna()` | Elimina filas con nulos |
| `.fillna(valor)` | Reemplaza nulos con un valor |

````python
# Valores nulos por columna
df.isnull().sum()
````

### 3.2 Filas duplicadas

| Funcion | Que hace |
|---------|----------|
| `.duplicated().sum()` | Cuenta filas duplicadas |
| `.drop_duplicates()` | Elimina filas duplicadas |

````python
# Duplicados
print(f'Filas duplicadas: {df.duplicated().sum()}')

# Eliminar duplicados si existen
df = df.drop_duplicates()
print(f'Despues de limpiar: {df.shape[0]} registros')
````

### 3.3 Filtros booleanos

Filtrar filas que cumplan una condicion.

```python
df[condicion]                    # una condicion
df[(condicion1) & (condicion2)]  # ambas condiciones (AND)
df[(condicion1) | (condicion2)]  # alguna condicion (OR)
```

````python
# Solo plantas con utilidad positiva
df_util = df[df['utilidad'] > 0]
print(f'Registros con utilidad positiva: {len(df_util)}')
````

````python
# Planta Sur Y producto Alpha
df_sur_alpha = df[(df['planta'] == 'Planta Sur') & (df['producto'] == 'Alpha')]
print(f'Registros Planta Sur + Alpha: {len(df_sur_alpha)}')
df_sur_alpha.head()
````

### 3.4 Agrupaciones (groupby)

Agrupar datos y calcular metricas por grupo.

```python
df.groupby('columna')['columna_num'].metodo()
```

````python
# Revenue promedio por planta
df.groupby('planta')['revenue'].mean()
````

````python
# Agregacion multiple: suma, promedio, maximo
df.groupby(['planta', 'producto']).agg({
    'revenue': 'sum',
    'utilidad': 'mean',
    'backlog': 'max'
})
````

### 3.5 Pivot Tables

Crear tablas dinamicas (similar a Excel).

```python
df.pivot_table(values='columna_num', index='fila', columns='columna', aggfunc='metodo')
```

````python
# Revenue promedio: planta (filas) x producto (columnas)
pivot = df.pivot_table(
    values='revenue',
    index='planta',
    columns='producto',
    aggfunc='mean'
)
pivot.round(0)
````

### 3.6 Columnas derivadas

Crear nuevas columnas a partir de las existentes.

````python
# Margen porcentual
df['margen_pct'] = (df['revenue'] - df['costo_produccion']) / df['revenue'] * 100

# Ratio produccion / demanda
df['ratio_pd'] = df['produccion'] / df['demanda']

df[['planta', 'producto', 'revenue', 'margen_pct', 'ratio_pd']].head()
````

## Paso 4: Estadistica Descriptiva

Resumir la distribucion de los datos con medidas numericas.

| Medida | Funcion | Que mide |
|--------|---------|----------|
| Promedio | `.mean()` | Centro de la distribucion |
| Mediana | `.median()` | Valor central (resistente a outliers) |
| Desviacion estandar | `.std()` | Dispersion de los datos |
| Correlacion | `.corr()` | Relacion lineal entre variables |
| Frecuencia | `.value_counts()` | Conteo por categoria |

````python
# Medidas centrales y dispersion
vars_numericas = ['revenue', 'utilidad', 'produccion', 'demanda', 'inventario']

for col in vars_numericas:
    print(f'{col:20s} | Media: {df[col].mean():>12,.0f} | Mediana: {df[col].median():>12,.0f} | Std: {df[col].std():>12,.0f}')
````

## Matriz de Correlación

### Para qué sirve:
- Identificar qué variables están relacionadas entre sí
- Detectar variables redundantes (alta correlación)
- Seleccionar features para modelos de ML
- Entender patrones en los datos

### Valores:
| Valor | Significado |
|-------|-------------|
| **1** | Correlación positiva perfecta (suben juntas) |
| **0** | Sin correlación |
| **-1** | Correlación negativa perfecta (una sube, otra baja) |

````python
# Matriz de correlacion
num_cols = df.select_dtypes(include='number').columns.tolist()
corr = df[num_cols].corr()

# Correlaciones mas fuertes con utilidad
corr['utilidad'].sort_values(ascending=False)
````

````python
# Frecuencias
print('=== Registros por Planta ===')
print(df['planta'].value_counts())
print()
print('=== Registros por Producto ===')
print(df['producto'].value_counts())
````

## Paso 5: Visualizacion de Datos

Crear graficos para comunicar hallazgos.

| Tipo de grafico | Cuando usar |
|-----------------|-------------|
| `hist` | Distribucion de una variable numerica |

### 5.1 Histograma - Distribucion de Revenue

````python
df['revenue'].hist(bins=30, figsize=(10, 6), edgecolor='black', color='#2e4165')
plt.title('Distribucion de Revenue', fontsize=14)
plt.xlabel('Revenue (CLP)')
plt.ylabel('Frecuencia')
plt.tight_layout()
plt.show()
````

### 5.2 Grafico de Barras - Revenue por Planta

````python
revenue_planta = df.groupby('planta')['revenue'].sum().sort_values(ascending=True)

revenue_planta.plot(kind='barh', figsize=(10, 6), color='#2e4165', edgecolor='black')
plt.title('Revenue Total por Planta', fontsize=14)
plt.xlabel('Revenue (CLP)')
plt.ylabel('')
plt.tight_layout()
plt.show()
````

### 5.3 Scatter Plot - Produccion vs Utilidad

````python
df.plot.scatter(x='produccion', y='utilidad', figsize=(10, 6), alpha=0.5, color='#c73526')
plt.title('Produccion vs Utilidad', fontsize=14)
plt.xlabel('Produccion (unidades)')
plt.ylabel('Utilidad (CLP)')
plt.tight_layout()
plt.show()
````

### 5.4 Heatmap de Correlaciones

````python
# Seleccionar variables clave para el heatmap
vars_heatmap = ['produccion', 'demanda', 'ventas', 'inventario', 'revenue',
                 'utilidad', 'backlog', 'satisfaccion_cliente', 'precio_unitario']

plt.figure(figsize=(12, 8))
sns.heatmap(df[vars_heatmap].corr(), annot=True, cmap='coolwarm', fmt='.2f',
            center=0, square=True, linewidths=0.5)
plt.title('Matriz de Correlaciones', fontsize=14)
plt.tight_layout()
plt.show()
````

### 5.5 Grafico de Lineas - Evolucion Temporal

````python
df['fecha'] = pd.to_datetime(df['fecha'])
evolucion = df.groupby('fecha')[['revenue', 'utilidad']].sum()

evolucion.plot(figsize=(12, 6))
plt.title('Evolucion Mensual: Revenue vs Utilidad', fontsize=14)
plt.xlabel('Fecha')
plt.ylabel('CLP')
plt.legend(['Revenue', 'Utilidad'])
plt.tight_layout()
plt.show()
````

### 5.6 Boxplot - Distribucion de Utilidad por Planta

````python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='planta', y='utilidad', palette='Set2')
plt.title('Distribucion de Utilidad por Planta', fontsize=14)
plt.xlabel('Planta')
plt.ylabel('Utilidad (CLP)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

## Paso 6: Desafio Final - Dashboard HTML (uso de opencode)
### Objetivo

Crear un archivo `dashboard.html` con **HTML/CSS puro** (sin JavaScript) que muestre un resumen ejecutivo de la empresa ManufacTech SpA.

### Requisitos del dashboard

1. **Header** con nombre de la empresa (encabezado en Markdown)
2. **4 KPI Cards**: Revenue total, Utilidad promedio, Satisfaccion, Backlog promedio
3. **Tabla resumen por planta** (5 filas)
4. **Tabla top 5 productos por margen**
5. **Footer** con Departamento de Ingenieria Industrial

### Paso 6.1 - Calcular los datos para el dashboard

````python
# === KPIs para el dashboard ===
kpi = {
    'revenue_total': df['revenue'].sum(),
    'utilidad_prom': df['utilidad'].mean(),
    'satisfaccion': df['satisfaccion_cliente'].mean(),
    'backlog_prom': df['backlog'].mean(),
}

print('KPIs calculados:')
for k, v in kpi.items():
    print(f'  {k}: {v:,.2f}')
````

````python
# === Resumen por planta ===
resumen_planta = df.groupby('planta').agg({
    'produccion': 'sum',
    'demanda': 'sum',
    'ventas': 'sum',
    'revenue': 'sum',
    'utilidad': 'sum',
    'satisfaccion_cliente': 'mean'
}).round(0)

resumen_planta
````

````python
# === Top 5 productos por margen ===
top_productos = df.groupby('producto').agg({
    'revenue': 'sum',
    'costo_produccion': 'sum',
    'utilidad': 'sum'
}).round(0)

top_productos['margen_pct'] = ((top_productos['revenue'] - top_productos['costo_produccion']) / top_productos['revenue'] * 100).round(1)
top_productos = top_productos.sort_values('margen_pct', ascending=False).head(5)

top_productos
````

````python
# === Evolucion mensual ===
df['fecha'] = pd.to_datetime(df['fecha'])
evolucion = df.groupby('fecha').agg({
    'revenue': 'sum',
    'utilidad': 'sum',
    'produccion': 'sum'
}).reset_index()

evolucion.head()
````

### Paso 6.2 - Generar el dashboard HTML

Ejecuta la siguiente celda para generar `dashboard_ejecutivo.html`.

````python
def generar_dashboard(df, kpi, resumen_plantas, top_prods, evol, ruta='dashboard_ejecutivo.html'):
    """Genera un dashboard HTML estatico con CSS grid."""
    
    # Formatear numeros
    def fmt(n, dec=0):
        if dec == 0:
            return f'{n:,.0f}'
        return f'{n:,.{dec}f}'
    
    # Filas de plantas
    filas_plantas = ''
    for nombre, row in resumen_plantas.iterrows():
        filas_plantas += f'''<tr>
            <td>{nombre}</td>
            <td>{fmt(row['produccion'])}</td>
            <td>{fmt(row['demanda'])}</td>
            <td>{fmt(row['ventas'])}</td>
            <td>${fmt(row['revenue'])}</td>
            <td>${fmt(row['utilidad'])}</td>
            <td>{fmt(row['satisfaccion_cliente'], 1)}</td>
        </tr>\n'''
    
    # Filas de productos
    filas_productos = ''
    for nombre, row in top_prods.iterrows():
        filas_productos += f'''<tr>
            <td>{nombre}</td>
            <td>${fmt(row['revenue'])}</td>
            <td>${fmt(row['costo_produccion'])}</td>
            <td>${fmt(row['utilidad'])}</td>
            <td>{fmt(row['margen_pct'], 1)}%</td>
        </tr>\n'''
    
    # Filas de evolucion
    filas_evol = ''
    for _, row in evol.iterrows():
        fecha_str = row['fecha'].strftime('%Y-%m')
        filas_evol += f'''<tr>
            <td>{fecha_str}</td>
            <td>${fmt(row['revenue'])}</td>
            <td>${fmt(row['utilidad'])}</td>
        </tr>\n'''
    
    html = f'''<!DOCTYPE html>
<html lang="es">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>ManufacTech SpA - Resumen Ejecutivo</title>
    <style>
        * {{ margin: 0; padding: 0; box-sizing: border-box; }}
        body {{ font-family: 'Helvetica', 'Arial', sans-serif; background: #f5f7fa; color: #2d2d2d; }}
        .header {{ background: #2e4165; color: white; padding: 24px 32px; text-align: center; }}
        .header h1 {{ font-size: 24pt; margin-bottom: 4px; }}
        .header p {{ font-size: 11pt; color: rgba(255,255,255,0.7); }}
        .container {{ max-width: 1100px; margin: 0 auto; padding: 24px; }}
        .kpi-grid {{ display: grid; grid-template-columns: repeat(4, 1fr); gap: 16px; margin-bottom: 32px; }}
        .kpi-card {{ background: white; border-radius: 8px; padding: 20px; text-align: center; box-shadow: 0 2px 8px rgba(0,0,0,0.08); border-top: 4px solid #2e4165; }}
        .kpi-card.red {{ border-top-color: #c73526; }}
        .kpi-card.gold {{ border-top-color: #956e3c; }}
        .kpi-card.green {{ border-top-color: #2e7d32; }}
        .kpi-value {{ font-size: 22pt; font-weight: bold; color: #2e4165; margin: 8px 0; }}
        .kpi-card.red .kpi-value {{ color: #c73526; }}
        .kpi-card.gold .kpi-value {{ color: #956e3c; }}
        .kpi-card.green .kpi-value {{ color: #2e7d32; }}
        .kpi-label {{ font-size: 9pt; color: #888; text-transform: uppercase; letter-spacing: 1px; }}
        .section {{ background: white; border-radius: 8px; padding: 24px; margin-bottom: 24px; box-shadow: 0 2px 8px rgba(0,0,0,0.08); }}
        .section h2 {{ font-size: 14pt; color: #2e4165; border-bottom: 2px solid #2e4165; padding-bottom: 8px; margin-bottom: 16px; }}
        table {{ width: 100%; border-collapse: collapse; font-size: 10pt; }}
        th {{ background: #2e4165; color: white; padding: 10px 12px; text-align: left; }}
        td {{ padding: 8px 12px; border-bottom: 1px solid #e0e0e0; }}
        tr:nth-child(even) {{ background: #f8f9fa; }}
        .footer {{ text-align: center; padding: 20px; color: #999; font-size: 9pt; border-top: 1px solid #e0e0e0; margin-top: 16px; }}
    </style>
</head>
<body>
    <div class="header">
        <h1>ManufacTech SpA</h1>
        <p>Resumen Ejecutivo - Periodo 2023-2025</p>
    </div>
    <div class="container">
        <div class="kpi-grid">
            <div class="kpi-card">
                <div class="kpi-label">Revenue Total</div>
                <div class="kpi-value">${fmt(kpi['revenue_total'])}</div>
                <div class="kpi-label">CLP</div>
            </div>
            <div class="kpi-card red">
                <div class="kpi-label">Utilidad Promedio</div>
                <div class="kpi-value">${fmt(kpi['utilidad_prom'])}</div>
                <div class="kpi-label">CLP / mes</div>
            </div>
            <div class="kpi-card gold">
                <div class="kpi-label">Satisfaccion</div>
                <div class="kpi-value">{fmt(kpi['satisfaccion'], 1)}</div>
                <div class="kpi-label">/ 100</div>
            </div>
            <div class="kpi-card green">
                <div class="kpi-label">Backlog Promedio</div>
                <div class="kpi-value">{fmt(kpi['backlog_prom'], 1)}</div>
                <div class="kpi-label">unidades</div>
            </div>
        </div>
        <div class="section">
            <h2>Resumen por Planta</h2>
            <table>
                <tr><th>Planta</th><th>Produccion</th><th>Demanda</th><th>Ventas</th><th>Revenue</th><th>Utilidad</th><th>Satisfaccion</th></tr>
                {filas_plantas}
            </table>
        </div>
        <div class="section">
            <h2>Top 5 Productos por Margen</h2>
            <table>
                <tr><th>Producto</th><th>Revenue</th><th>Costo Prod.</th><th>Utilidad</th><th>Margen</th></tr>
                {filas_productos}
            </table>
        </div>
        <div class="section">
            <h2>Evolucion Mensual</h2>
            <table>
                <tr><th>Mes</th><th>Revenue</th><th>Utilidad</th></tr>
                {filas_evol}
            </table>
        </div>
    </div>
    <div class="footer">
        Departamento de Ingenieria Industrial - Modelacion de Sistemas - Agosto 2026
    </div>
</body>
</html>'''
    
    with open(ruta, 'w', encoding='utf-8') as f:
        f.write(html)
    
    print(f'Dashboard generado: {ruta}')

generar_dashboard(df, kpi, resumen_planta, top_productos, evolucion)
````
